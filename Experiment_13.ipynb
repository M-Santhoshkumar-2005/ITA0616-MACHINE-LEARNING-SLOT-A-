```python
import numpy as np
import pandas as pd


dataset = [
    [2015, 68000, 1.5, 1, 5.2],
    [2016, 59000, 1.5, 1, 5.8],
    [2017, 48000, 2.0, 1, 6.9],
    [2018, 42000, 2.0, 2, 8.1],
    [2019, 35000, 2.5, 2, 9.4],
    [2020, 28000, 2.5, 2, 10.8],
    [2021, 22000, 3.0, 3, 12.6],
    [2022, 16000, 3.0, 3, 14.5],
    [2023, 11000, 3.5, 4, 17.2],
    [2024, 6500, 3.5, 4, 19.8],
    [2016, 64000, 1.2, 1, 4.9],
    [2018, 46000, 1.8, 2, 7.6],
    [2020, 31000, 2.2, 2, 10.2],
    [2021, 24500, 2.8, 3, 12.1],
    [2022, 18000, 3.2, 3, 14.0],
    [2023, 9500, 3.8, 4, 18.0],
]

df = pd.DataFrame(
    dataset,
    columns=["Year", "Mileage_Km", "Engine_L", "Owners", "Price_Lakhs"],
)

X = df[["Year", "Mileage_Km", "Engine_L", "Owners"]].values.astype(float)
y = df["Price_Lakhs"].values.astype(float)

mean = X.mean(axis=0)
std = np.where(X.std(axis=0) == 0, 1.0, X.std(axis=0))
X_scaled = (X - mean) / std

X_design = np.column_stack([np.ones(len(X_scaled)), X_scaled])
theta = np.linalg.pinv(X_design) @ y

y_pred = X_design @ theta
mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)
mae = np.mean(np.abs(y - y_pred))
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1.0 - (ss_res / ss_tot)

print(f"MSE: {mse:.4f} | RMSE: {rmse:.4f} | MAE: {mae:.4f} | R2: {r2:.4f}\n")

for i, row in enumerate(X):
    print(
        f"Car {i + 1:2d} | Year={int(row[0])}, Km={int(row[1]):5d}, "
        f"Eng={row[2]:.1f}L, Own={int(row[3])} | "
        f"Actual={y[i]:5.2f}L | Pred={y_pred[i]:5.2f}L"
    )

new_cars = np.array([
    [2019, 36000, 2.0, 2],
    [2023, 10000, 3.5, 4],
])
new_scaled = (new_cars - mean) / std
new_design = np.column_stack([np.ones(len(new_scaled)), new_scaled])
new_preds = new_design @ theta

print("\nNew Car Predictions:")
for idx, (car, pr) in enumerate(zip(new_cars, new_preds), start=1):
    print(
        f"New Car {idx} | Year={int(car[0])}, Km={int(car[1]):5d}, "
        f"Eng={car[2]:.1f}L, Own={int(car[3])} | Predicted Price={pr:5.2f} Lakhs"
    )
```
