```python
import numpy as np
import pandas as pd


dataset = [
    [42.0],
    [45.0],
    [44.0],
    [47.0],
    [43.0],
    [46.0],
    [48.0],
    [45.5],
    [72.0],
    [75.0],
    [70.0],
    [78.0],
    [74.0],
    [76.0],
    [73.0],
    [77.0],
]

df = pd.DataFrame(dataset, columns=["Exam_Score"])
X = df["Exam_Score"].values.astype(float)
n = len(X)

pi = np.array([0.5, 0.5])
mu = np.array([50.0, 68.0])
sigma = np.array([8.0, 8.0])


def gaussian_pdf(x, mean, std):
    var = np.maximum(std ** 2, 1e-6)
    coeff = 1.0 / np.sqrt(2.0 * np.pi * var)
    exponent = -((x - mean) ** 2) / (2.0 * var)
    return coeff * np.exp(exponent)


max_iter = 200
tol = 1e-6
prev_ll = -np.inf

for iteration in range(1, max_iter + 1):
    weighted_pdf = np.column_stack([
        pi[k] * gaussian_pdf(X, mu[k], sigma[k]) for k in range(2)
    ])
    total_prob = np.sum(weighted_pdf, axis=1, keepdims=True)
    total_prob = np.maximum(total_prob, 1e-15)
    gamma = weighted_pdf / total_prob

    Nk = np.sum(gamma, axis=0)
    pi = Nk / n
    for k in range(2):
        Nk_safe = max(Nk[k], 1e-12)
        mu[k] = np.sum(gamma[:, k] * X) / Nk_safe
        var_k = np.sum(gamma[:, k] * ((X - mu[k]) ** 2)) / Nk_safe
        sigma[k] = np.sqrt(max(var_k, 1e-6))

    ll = np.sum(np.log(np.maximum(np.sum(
        np.column_stack([pi[k] * gaussian_pdf(X, mu[k], sigma[k]) for k in range(2)]),
        axis=1
    ), 1e-15)))

    if abs(ll - prev_ll) < tol:
        break
    prev_ll = ll

weighted_pdf = np.column_stack([
    pi[k] * gaussian_pdf(X, mu[k], sigma[k]) for k in range(2)
])
gamma = weighted_pdf / np.maximum(np.sum(weighted_pdf, axis=1, keepdims=True), 1e-15)
clusters = np.argmax(gamma, axis=1)

print(f"Converged in {iteration} iterations | Log-Likelihood: {prev_ll:.4f}\n")
for k in range(2):
    print(f"Cluster {k + 1} | Weight (pi): {pi[k]:.4f} | Mean (mu): {mu[k]:.2f} | Std Dev (sigma): {sigma[k]:.2f}")

print("\nSample Assignments:")
for i, val in enumerate(X):
    print(
        f"Sample {i + 1:2d} | Score={val:5.1f} | "
        f"Resp_C1={gamma[i, 0]:.4f}, Resp_C2={gamma[i, 1]:.4f} | Cluster={clusters[i] + 1}"
    )
```
