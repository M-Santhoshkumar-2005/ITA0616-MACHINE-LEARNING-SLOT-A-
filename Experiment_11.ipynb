```python
import numpy as np
import pandas as pd


dataset = [
    [25000, 12000, 72, 1, 0],
    [30000, 15000, 68, 2, 0],
    [35000, 14000, 75, 1, 0],
    [28000, 16000, 62, 3, 0],
    [40000, 18000, 70, 2, 0],
    [52000, 15000, 82, 4, 1],
    [60000, 14000, 85, 5, 1],
    [55000, 16000, 80, 4, 1],
    [65000, 12000, 88, 6, 1],
    [58000, 13000, 84, 5, 1],
    [85000, 10000, 95, 8, 2],
    [95000, 9000, 98, 10, 2],
    [90000, 11000, 92, 9, 2],
    [105000, 8000, 97, 12, 2],
    [110000, 7500, 99, 15, 2],
    [48000, 17000, 78, 3, 1],
    [72000, 11500, 89, 7, 1],
    [88000, 9500, 94, 9, 2],
]

df = pd.DataFrame(
    dataset,
    columns=["Annual_Income", "Monthly_Debt", "Payment_History_Pct", "Credit_History_Yrs", "Credit_Score"],
)

X = df[["Annual_Income", "Monthly_Debt", "Payment_History_Pct", "Credit_History_Yrs"]].values.astype(float)
y = df["Credit_Score"].values.astype(int)
num_classes = len(np.unique(y))

mean = X.mean(axis=0)
std = X.std(axis=0)
X_scaled = (X - mean) / std

Y_onehot = np.eye(num_classes)[y]


def softmax(z):
    z_shifted = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(z_shifted)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)


rng = np.random.default_rng(42)
W = rng.normal(0.0, 0.1, size=(X_scaled.shape[1], num_classes))
b = np.zeros((1, num_classes))
lr = 0.2
epochs = 2000

for _ in range(epochs):
    probs = softmax(X_scaled @ W + b)
    m = len(y)
    dW = (X_scaled.T @ (probs - Y_onehot)) / m
    db = np.mean(probs - Y_onehot, axis=0, keepdims=True)
    W -= lr * dW
    b -= lr * db

probs = softmax(X_scaled @ W + b)
preds = np.argmax(probs, axis=1)
accuracy = np.mean(preds == y) * 100

score_names = {0: "Poor", 1: "Standard", 2: "Good"}
print(f"Training Accuracy: {accuracy:.2f}%\n")

for i, row in enumerate(X):
    print(
        f"Sample {i + 1:2d} | Income={int(row[0]):6d}, Debt={int(row[1]):5d}, "
        f"PayHist={int(row[2]):2d}%, HistYrs={int(row[3]):2d} | "
        f"Actual={score_names[y[i]]:8s} | Pred={score_names[preds[i]]:8s}"
    )

new_applicants = np.array([
    [32000, 15000, 66, 2],
    [62000, 12500, 86, 6],
    [98000, 8500, 96, 11],
])
new_scaled = (new_applicants - mean) / std
new_probs = softmax(new_scaled @ W + b)
new_preds = np.argmax(new_probs, axis=1)

print("\nNew Applicant Predictions:")
for idx, (app, pr, p_row) in enumerate(zip(new_applicants, new_preds, new_probs), start=1):
    print(
        f"Applicant {idx} | Income={int(app[0]):6d}, Debt={int(app[1]):5d}, "
        f"PayHist={int(app[2]):2d}%, HistYrs={int(app[3]):2d} | "
        f"Pred={score_names[pr]:8s} | Probs={np.round(p_row, 3)}"
    )
```
