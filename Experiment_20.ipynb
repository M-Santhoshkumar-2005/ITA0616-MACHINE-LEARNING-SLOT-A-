```python
import numpy as np
import pandas as pd


dataset = [
    [1, 1000, 120000],
    [2, 1200, 132000],
    [3, 1100, 128000],
    [4, 1500, 155000],
    [5, 1800, 178000],
    [6, 2200, 205000],
    [7, 2500, 232000],
    [8, 2400, 228000],
    [9, 2900, 265000],
    [10, 3400, 305000],
    [11, 4200, 368000],
    [12, 4800, 420000],
]

df = pd.DataFrame(dataset, columns=["Month_Index", "Ad_Spend_INR", "Sales_INR"])

t = df["Month_Index"].values.astype(float)
spend = df["Ad_Spend_INR"].values.astype(float)
y = df["Sales_INR"].values.astype(float)

X_design = np.column_stack([
    np.ones_like(t),
    t,
    spend,
    np.sin(2 * np.pi * t / 12.0),
    np.cos(2 * np.pi * t / 12.0),
])

theta = np.linalg.pinv(X_design) @ y
y_pred = X_design @ theta

mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)
mae = np.mean(np.abs(y - y_pred))
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1.0 - (ss_res / ss_tot)

print(f"MSE: {mse:.2f} | RMSE: {rmse:.2f} | MAE: {mae:.2f} | R2: {r2:.4f}\n")

for i in range(len(t)):
    print(
        f"Month {int(t[i]):2d} | Ad_Spend=₹{int(spend[i]):5d} | "
        f"Actual=₹{y[i]:9.2f} | Fitted=₹{y_pred[i]:9.2f}"
    )

future_months = np.array([13, 14, 15, 16])
future_spend = np.array([3600, 3800, 4100, 4500], dtype=float)

X_future = np.column_stack([
    np.ones_like(future_months),
    future_months,
    future_spend,
    np.sin(2 * np.pi * future_months / 12.0),
    np.cos(2 * np.pi * future_months / 12.0),
])
future_preds = X_future @ theta

print("\nFuture Sales Forecast:")
for m_idx, sp, pr in zip(future_months, future_spend, future_preds):
    print(f"Future Month {int(m_idx):2d} | Planned Ad_Spend=₹{int(sp):5d} | Forecast Sales=₹{pr:9.2f}")
```
