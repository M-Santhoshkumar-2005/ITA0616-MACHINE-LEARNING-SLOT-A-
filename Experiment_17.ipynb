```python
import numpy as np
import pandas as pd


dataset = [
    [128, 6, 5000, 12, 11999],
    [128, 6, 5000, 13, 12999],
    [256, 8, 5000, 50, 16999],
    [256, 8, 5000, 64, 18999],
    [128, 8, 5000, 48, 15499],
    [64, 4, 4500, 8, 8999],
    [64, 4, 5000, 13, 9999],
    [128, 4, 5000, 13, 10499],
    [256, 12, 5000, 64, 24999],
    [512, 12, 5000, 108, 32999],
    [256, 8, 4500, 50, 15999],
    [128, 6, 4500, 12, 11499],
    [512, 16, 5500, 108, 39999],
    [256, 12, 5500, 64, 26999],
    [64, 6, 5000, 13, 10999],
    [128, 8, 5500, 50, 16499],
]

df = pd.DataFrame(
    dataset,
    columns=["Storage_GB", "RAM_GB", "Battery_mAh", "Camera_MP", "Price_INR"],
)

X = df[["Storage_GB", "RAM_GB", "Battery_mAh", "Camera_MP"]].values.astype(float)
y = df["Price_INR"].values.astype(float)

mean = X.mean(axis=0)
std = np.where(X.std(axis=0) == 0, 1.0, X.std(axis=0))
X_scaled = (X - mean) / std

X_design = np.column_stack([np.ones(len(X_scaled)), X_scaled])
theta = np.linalg.pinv(X_design) @ y

y_pred = X_design @ theta
mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)
mae = np.mean(np.abs(y - y_pred))
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1.0 - (ss_res / ss_tot)

print(f"MSE: {mse:.2f} | RMSE: {rmse:.2f} | MAE: {mae:.2f} | R2: {r2:.4f}\n")

for i, row in enumerate(X):
    print(
        f"Phone {i + 1:2d} | {int(row[0]):3d}GB, {int(row[1]):2d}GB RAM, "
        f"{int(row[2]):4d}mAh, {int(row[3]):3d}MP | "
        f"Actual=₹{y[i]:8.2f} | Pred=₹{y_pred[i]:8.2f}"
    )

new_phones = np.array([
    [128, 8, 5000, 50],
    [256, 12, 5000, 64],
])
new_scaled = (new_phones - mean) / std
new_design = np.column_stack([np.ones(len(new_scaled)), new_scaled])
new_preds = new_design @ theta

print("\nNew Mobile Price Predictions:")
for idx, (ph, pr) in enumerate(zip(new_phones, new_preds), start=1):
    print(
        f"New Phone {idx} | {int(ph[0]):3d}GB, {int(ph[1]):2d}GB RAM, "
        f"{int(ph[2]):4d}mAh, {int(ph[3]):3d}MP | Predicted Price=₹{pr:8.2f}"
    )
```
