```python
import numpy as np
import pandas as pd


dataset = [
    [2.0, 45.0, 0],
    [2.5, 50.0, 0],
    [3.0, 48.0, 0],
    [3.5, 55.0, 0],
    [4.0, 52.0, 0],
    [4.5, 60.0, 0],
    [5.0, 58.0, 0],
    [5.5, 65.0, 1],
    [6.0, 68.0, 1],
    [6.5, 72.0, 1],
    [7.0, 75.0, 1],
    [7.5, 80.0, 1],
    [8.0, 82.0, 1],
    [8.5, 85.0, 1],
    [9.0, 90.0, 1],
    [9.5, 92.0, 1],
]

df = pd.DataFrame(dataset, columns=["Study_Hours", "Attendance_Pct", "Pass"])

X = df[["Study_Hours", "Attendance_Pct"]].values.astype(float)
y = df["Pass"].values.astype(float).reshape(-1, 1)

mean = X.mean(axis=0)
std = X.std(axis=0)
X_scaled = (X - mean) / std


def sigmoid(z):
    z_clipped = np.clip(z, -500, 500)
    return 1.0 / (1.0 + np.exp(-z_clipped))


def compute_loss(y_true, y_pred):
    eps = 1e-15
    y_pred = np.clip(y_pred, eps, 1.0 - eps)
    return -np.mean(y_true * np.log(y_pred) + (1.0 - y_true) * np.log(1.0 - y_pred))


rng = np.random.default_rng(42)
weights = rng.normal(0.0, 0.1, size=(X_scaled.shape[1], 1))
bias = 0.0
lr = 0.1
epochs = 3000

for epoch in range(1, epochs + 1):
    z = X_scaled @ weights + bias
    y_pred = sigmoid(z)
    m = len(y)
    dw = (X_scaled.T @ (y_pred - y)) / m
    db = np.mean(y_pred - y)
    weights -= lr * dw
    bias -= lr * db

probs = sigmoid(X_scaled @ weights + bias)
preds = (probs >= 0.5).astype(int)
accuracy = np.mean(preds.flatten() == y.flatten().astype(int)) * 100

print(f"Final Loss: {compute_loss(y, probs):.4f}")
print(f"Training Accuracy: {accuracy:.2f}%\n")

for i, row in enumerate(X):
    print(
        f"Sample {i + 1:2d} | Study_Hours={row[0]:4.1f}, Attendance={row[1]:4.1f}% | "
        f"Actual={int(y[i, 0])} | Pred={int(preds[i, 0])} (prob={probs[i, 0]:.4f})"
    )

new_students = np.array([
    [3.8, 54.0],
    [6.2, 70.0],
])
new_scaled = (new_students - mean) / std
new_probs = sigmoid(new_scaled @ weights + bias)
new_preds = (new_probs >= 0.5).astype(int)

print("\nNew Predictions:")
for idx, (st, p_val, pr) in enumerate(zip(new_students, new_probs.flatten(), new_preds.flatten()), start=1):
    print(f"New {idx} | Study_Hours={st[0]:4.1f}, Attendance={st[1]:4.1f}% | Pred={pr} (prob={p_val:.4f})")
```
