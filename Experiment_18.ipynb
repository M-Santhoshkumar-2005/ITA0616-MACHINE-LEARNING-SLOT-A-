```python
import numpy as np
import pandas as pd


dataset = [
    [5.1, 3.5, 1.4, 0.2, 0],
    [4.9, 3.0, 1.4, 0.2, 0],
    [4.7, 3.2, 1.3, 0.2, 0],
    [4.6, 3.1, 1.5, 0.2, 0],
    [5.0, 3.6, 1.4, 0.2, 0],
    [5.4, 3.9, 1.7, 0.4, 0],
    [4.6, 3.4, 1.4, 0.3, 0],
    [5.0, 3.4, 1.5, 0.2, 0],
    [7.0, 3.2, 4.7, 1.4, 1],
    [6.4, 3.2, 4.5, 1.5, 1],
    [6.9, 3.1, 4.9, 1.5, 1],
    [5.5, 2.3, 4.0, 1.3, 1],
    [6.5, 2.8, 4.6, 1.5, 1],
    [5.7, 2.8, 4.5, 1.3, 1],
    [6.3, 3.3, 6.0, 2.5, 1],
    [7.1, 3.0, 5.9, 2.1, 1],
    [6.5, 3.0, 5.8, 2.2, 1],
    [7.6, 3.0, 6.6, 2.1, 1],
]

df = pd.DataFrame(
    dataset,
    columns=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm", "Target"],
)

X = df[["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]].values.astype(float)
y = df["Target"].values.astype(int)

test_idx = [3, 6, 11, 14, 16]
train_mask = np.ones(len(X), dtype=bool)
train_mask[test_idx] = False

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_idx], y[test_idx]

mean = X_train.mean(axis=0)
std = np.where(X_train.std(axis=0) == 0, 1.0, X_train.std(axis=0))
X_tr_s = (X_train - mean) / std
X_te_s = (X_test - mean) / std

rng = np.random.default_rng(42)
weights = rng.normal(0.0, 0.1, size=X_tr_s.shape[1])
bias = 0.0
lr = 0.1
epochs = 200

for _ in range(epochs):
    errors = 0
    for x_i, y_i in zip(X_tr_s, y_train):
        z = np.dot(x_i, weights) + bias
        y_pred = 1 if z >= 0.0 else 0
        err = y_i - y_pred
        if err != 0:
            weights += lr * err * x_i
            bias += lr * err
            errors += 1
    if errors == 0:
        break

test_preds = np.where((X_te_s @ weights + bias) >= 0.0, 1, 0)
accuracy = np.mean(test_preds == y_test) * 100

label_map = {0: "Iris-setosa", 1: "Non-setosa (Versicolor/Virginica)"}
print(f"Test Accuracy: {accuracy:.2f}%\n")

for idx, row, actual, pred in zip(test_idx, X_test, y_test, test_preds):
    print(
        f"Row {idx + 1:2d} | SL={row[0]:.1f}, SW={row[1]:.1f}, PL={row[2]:.1f}, PW={row[3]:.1f} | "
        f"Actual={label_map[actual]} | Pred={label_map[pred]}"
    )

new_samples = np.array([
    [4.9, 3.1, 1.5, 0.1],
    [6.0, 2.9, 4.5, 1.5],
    [6.7, 3.0, 5.2, 2.3],
])
new_scaled = (new_samples - mean) / std
new_preds = np.where((new_scaled @ weights + bias) >= 0.0, 1, 0)

print("\nNew Sample Predictions:")
for i, (sample, pred) in enumerate(zip(new_samples, new_preds), start=1):
    print(
        f"New {i} | SL={sample[0]:.1f}, SW={sample[1]:.1f}, PL={sample[2]:.1f}, PW={sample[3]:.1f} | "
        f"Predicted={label_map[pred]}"
    )
```
