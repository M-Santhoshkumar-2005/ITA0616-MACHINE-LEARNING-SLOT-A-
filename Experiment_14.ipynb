```python
import numpy as np
import pandas as pd


dataset = [
    [950, 2, 1, 12, 42.0],
    [1100, 2, 1, 10, 48.5],
    [1250, 2, 2, 8, 56.0],
    [1400, 3, 2, 7, 64.5],
    [1600, 3, 2, 5, 75.0],
    [1800, 3, 3, 4, 86.0],
    [2000, 4, 3, 3, 98.5],
    [2250, 4, 3, 2, 112.0],
    [2500, 4, 4, 1, 128.0],
    [2800, 5, 4, 1, 145.0],
    [1050, 2, 1, 15, 44.0],
    [1500, 3, 2, 6, 69.0],
    [1750, 3, 2, 5, 81.5],
    [2100, 4, 3, 3, 104.0],
    [2400, 4, 3, 2, 120.0],
    [2700, 4, 4, 1, 138.0],
]

df = pd.DataFrame(
    dataset,
    columns=["Area_SqFt", "Bedrooms", "Bathrooms", "Age_Yrs", "Price_Lakhs"],
)

X = df[["Area_SqFt", "Bedrooms", "Bathrooms", "Age_Yrs"]].values.astype(float)
y = df["Price_Lakhs"].values.astype(float)

mean = X.mean(axis=0)
std = np.where(X.std(axis=0) == 0, 1.0, X.std(axis=0))
X_scaled = (X - mean) / std

X_design = np.column_stack([np.ones(len(X_scaled)), X_scaled])
theta = np.linalg.pinv(X_design) @ y

y_pred = X_design @ theta
mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)
mae = np.mean(np.abs(y - y_pred))
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1.0 - (ss_res / ss_tot)

print(f"MSE: {mse:.4f} | RMSE: {rmse:.4f} | MAE: {mae:.4f} | R2: {r2:.4f}\n")

for i, row in enumerate(X):
    print(
        f"House {i + 1:2d} | Area={int(row[0]):4d} sqft, Bed={int(row[1])}, "
        f"Bath={int(row[2])}, Age={int(row[3]):2d}y | "
        f"Actual=₹{y[i]:6.2f}L | Pred=₹{y_pred[i]:6.2f}L"
    )

new_houses = np.array([
    [1350, 2, 2, 8],
    [2200, 4, 3, 2],
])
new_scaled = (new_houses - mean) / std
new_design = np.column_stack([np.ones(len(new_scaled)), new_scaled])
new_preds = new_design @ theta

print("\nNew House Predictions:")
for idx, (h, pr) in enumerate(zip(new_houses, new_preds), start=1):
    print(
        f"New House {idx} | Area={int(h[0]):4d} sqft, Bed={int(h[1])}, "
        f"Bath={int(h[2])}, Age={int(h[3]):2d}y | Predicted Price=₹{pr:6.2f} Lakhs"
    )
```
