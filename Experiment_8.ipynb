```python
import numpy as np
import pandas as pd


dataset = [
    [2.0, 45.0, 52.0],
    [2.5, 50.0, 55.0],
    [3.0, 55.0, 58.0],
    [3.5, 60.0, 62.0],
    [4.0, 62.0, 65.0],
    [4.5, 68.0, 69.0],
    [5.0, 70.0, 72.0],
    [5.5, 75.0, 76.0],
    [6.0, 78.0, 79.0],
    [6.5, 82.0, 83.0],
    [7.0, 85.0, 86.0],
    [7.5, 88.0, 89.0],
    [8.0, 90.0, 92.0],
    [8.5, 92.0, 95.0],
    [9.0, 95.0, 97.0],
    [9.5, 96.0, 98.0],
]

df = pd.DataFrame(dataset, columns=["Study_Hours", "Attendance_Pct", "Final_Score"])

X = df[["Study_Hours", "Attendance_Pct"]].values.astype(float)
y = df["Final_Score"].values.astype(float)

mean = X.mean(axis=0)
std = X.std(axis=0)
X_scaled = (X - mean) / std

X_design = np.column_stack([np.ones(len(X_scaled)), X_scaled])

theta = np.linalg.pinv(X_design) @ y

y_pred = X_design @ theta
mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1.0 - (ss_res / ss_tot)

print(f"MSE: {mse:.4f} | RMSE: {rmse:.4f} | R2 Score: {r2:.4f}\n")

for i, row in enumerate(X):
    print(
        f"Sample {i + 1:2d} | Study_Hours={row[0]:4.1f}, Attendance={row[1]:4.1f}% | "
        f"Actual={y[i]:5.2f} | Pred={y_pred[i]:5.2f}"
    )

new_students = np.array([
    [4.2, 65.0],
    [7.2, 86.0],
])
new_scaled = (new_students - mean) / std
new_design = np.column_stack([np.ones(len(new_scaled)), new_scaled])
new_preds = new_design @ theta

print("\nNew Predictions:")
for idx, (st, pr) in enumerate(zip(new_students, new_preds), start=1):
    print(f"New {idx} | Study_Hours={st[0]:4.1f}, Attendance={st[1]:4.1f}% | Predicted Score={pr:5.2f}")
```
