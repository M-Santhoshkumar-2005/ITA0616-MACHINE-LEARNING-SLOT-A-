```python
import numpy as np
import pandas as pd


dataset = [
    [1.0, 12.0],
    [1.5, 13.5],
    [2.0, 15.0],
    [2.5, 17.5],
    [3.0, 21.0],
    [3.5, 26.0],
    [4.0, 33.0],
    [4.5, 42.0],
    [5.0, 53.0],
    [5.5, 66.0],
    [6.0, 81.0],
    [6.5, 98.0],
    [7.0, 117.0],
    [7.5, 138.0],
    [8.0, 161.0],
    [8.5, 186.0],
]

df = pd.DataFrame(dataset, columns=["Practice_Hours", "Problems_Solved"])

x = df["Practice_Hours"].values.astype(float)
y = df["Problems_Solved"].values.astype(float)

X_lin = np.column_stack([np.ones_like(x), x])
theta_lin = np.linalg.pinv(X_lin) @ y
y_pred_lin = X_lin @ theta_lin

X_poly = np.column_stack([np.ones_like(x), x, x ** 2])
theta_poly = np.linalg.pinv(X_poly) @ y
y_pred_poly = X_poly @ theta_poly


def regression_metrics(y_true, y_pred):
    mse = np.mean((y_true - y_pred) ** 2)
    rmse = np.sqrt(mse)
    mae = np.mean(np.abs(y_true - y_pred))
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    r2 = 1.0 - (ss_res / ss_tot)
    return mse, rmse, mae, r2


lin_mse, lin_rmse, lin_mae, lin_r2 = regression_metrics(y, y_pred_lin)
poly_mse, poly_rmse, poly_mae, poly_r2 = regression_metrics(y, y_pred_poly)

print("Linear Regression:")
print(f"  MSE: {lin_mse:.4f} | RMSE: {lin_rmse:.4f} | MAE: {lin_mae:.4f} | R2: {lin_r2:.4f}")

print("\nPolynomial Regression (degree=2):")
print(f"  MSE: {poly_mse:.4f} | RMSE: {poly_rmse:.4f} | MAE: {poly_mae:.4f} | R2: {poly_r2:.4f}\n")

for i in range(len(x)):
    print(
        f"Sample {i + 1:2d} | Hours={x[i]:4.1f} | Actual={y[i]:6.2f} | "
        f"Linear={y_pred_lin[i]:6.2f} | Poly={y_pred_poly[i]:6.2f}"
    )

test_hours = np.array([3.2, 6.8])
test_lin = np.column_stack([np.ones_like(test_hours), test_hours]) @ theta_lin
test_poly = np.column_stack([np.ones_like(test_hours), test_hours, test_hours ** 2]) @ theta_poly

print("\nNew Predictions:")
for h, p_lin, p_poly in zip(test_hours, test_lin, test_poly):
    print(f"Hours={h:4.1f} | Linear Pred={p_lin:6.2f} | Polynomial Pred={p_poly:6.2f}")
```
